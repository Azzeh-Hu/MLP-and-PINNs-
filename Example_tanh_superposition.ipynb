```python
import numpy as np
import matplotlib.pyplot as plt

def linear(x, w=0.4, b=0.0):
    return w*x + b

def single_tanh(x, w=1.5, b=0):
    return np.tanh(w*x + b)

def three_tanh_superim(x, weights, biases, amp):
    y = np.zeros_like(x)
    for w, b, v in zip (weights, biases, amp):
        y+= v * np.tanh(w * x + b) # this represents an output layer, without an activaion functiin but with only an amplification factor
    return y

def main():
    x = np.linspace(-3, 3, 300)
    y_linear = linear(x)
    y_one = single_tanh(x)

    weights = [3.0, 2.5, 4.0]
    biases = [3.0, 0.0, -4.0]      # shifts the bend left/right
    amp = [0.5, 0.6, 0.4]
    y_three = three_tanh_superim(x, weights , biases , amp)

    fig, ax = plt.subplots(figsize=(8,6))
    ax.axhline(0, color = "black" , linewidth = 0.6)
    ax.axvline(0, color = "black" , linewidth = 0.6)

    ax.plot(x, y_linear, label = " linear function $wx+b$",
            color = "grey")
    ax.plot(x, y_one, label = "one tnah neurone",
            color = "tab:blue")
    ax.plot(x, y_three, label= "three tanh nuerones superimposed",
             color = "tab:red")
    ax.set_xlabel("x")
    ax.set_ylabel("f(x)")
    ax.set_title("tnah neurone superposition")
    ax.legend(loc = "upper left", fontsize =9)
    ax.set_ylim(-2,2)

    fig.tight_layout()
    fig.savefig("tanh_superposition_png", dpi=150)
    print("saved plot")

if __name__== "__main__":
    main()





```
