```python



import numpy as np
def true_function(x1,x2):
    return np.sin(x1) + x1 * x2

def sigmoid(z):
    return 1.0/ (1 + np.exp(-z))

def tanh(z):
    return np.tanh(z)

def main():
    x1, x2 = 0.6, 0.9
    f_true = true_function(x1, x2)

    # pereceptron parameters
    w1, w2, b = -0.8 , 1.5, -0.3

    z = w1*x1 + w2*x2 + 1*b

    a_sigmoid = sigmoid(z)
    a_tanh = tanh(z)


    err_sigmoid = abs(f_true - a_sigmoid)
    err_tanh = abs(f_true - a_tanh)
    rel_err_sigmoid = err_sigmoid/f_true * 100
    rel_err_tanh= err_tanh/f_true * 100

    print(f" f({x1},{x2}) = {f_true:.3f}")
    print(f" pre-activation z = w1*x1 + w2*x2 + b = {z:.3f}\n")
    print("sigmoid function")
    print(f" a = sigmoid(z) = { a_sigmoid:.3f}")
    print(f" error = |{f_true:.3f} - {a_sigmoid:.3f}| = {err_sigmoid:.3f} ")
    print(f" relative error = {rel_err_sigmoid:.0f}%\n")

    print(f"tnah activation:")
    print(f" a_tanh(z) = {a_tanh:.3f}")
    print(f" error =  {f_true:.3f} -  {a_tanh:.3f} = {err_tanh:.3f}")
    print(f"relative error = {rel_err_tanh:.0f}%")


if __name__ == "__main__":
    main()
```
